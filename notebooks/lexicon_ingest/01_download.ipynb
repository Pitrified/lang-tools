# 01 - Download the raw OMW backbone

Stage A of the phase-5 initial build. Pulls OMW into the gitignored raw cache
under `data/_raw/lexicon/` and returns the manifest fragment that pins its exact
version. Run **rarely** - the download is large and the cache is reproducible.

Thin caller: all logic is in `ingestion.acquire`. Needs the `ingest` extra for
OMW (`uv sync --extra ingest`). The kaikki enrichment fetch was removed in phase
5.5: OMW glosses plus a CILI English fallback cover the gloss need, with no
CC-BY-SA source pulled in.


```python
from loguru import logger as lg

from lang_tools.lexicon.ingestion.acquire import download_omw
from lang_tools.params.lang_tools_params import get_lang_tools_params

LANGS = [
    "en",
    "pt",
    "es",
    "fr",
    "it",
]
data_fol = get_lang_tools_params().paths.data_fol
lg.info("01-download: langs={} -> data_fol={}", LANGS, data_fol)
data_fol
```

## Download OMW (the concept backbone)


```python
lg.info("Downloading OMW lexicons + CILI for {} (large; idempotent)...", LANGS)
omw_info = download_omw(LANGS, data_fol=data_fol)
lg.success("OMW + CILI download complete: {}", omw_info)
omw_info
```

Keep `omw_info` - pass it to `build_initial(..., extra_manifest=omw_info)` in
`02_transform` so the `_build.json` manifest pins the source version.
