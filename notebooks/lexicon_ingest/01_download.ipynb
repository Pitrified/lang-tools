# 01 - Download raw OMW + kaikki

Stage A of the phase-5 initial build. Pulls the external sources into the
gitignored raw cache under `data/_raw/lexicon/` and returns the manifest
fragments that pin their exact versions. Run **rarely** - the downloads are
large and the cache is reproducible.

Thin caller: all logic is in `ingestion.acquire`. Needs the `ingest` extra for
OMW (`uv sync --extra ingest`); the kaikki fetch is a plain HTTPS GET.

The kaikki dumps are big (hundreds of MB per language). Comment out
`fetch_kaikki` to build OMW-only first.


```python
from lang_tools.lexicon.ingestion.acquire import download_omw
from lang_tools.lexicon.ingestion.acquire import fetch_kaikki
from lang_tools.params.lang_tools_params import get_lang_tools_params

LANGS = [
    "en",
    "pt",
    "es",
    "fr",
    "it",
]
data_fol = get_lang_tools_params().paths.data_fol
data_fol
```

## Download OMW (the concept backbone)


```python
omw_info = download_omw(LANGS, data_fol=data_fol)
omw_info
```

## Fetch kaikki (enrichment, CC-BY-SA, license-isolated)


```python
kaikki_info = fetch_kaikki(LANGS, data_fol=data_fol)
kaikki_info
```

Keep `omw_info` / `kaikki_info` - pass them to `build_initial(..., extra_manifest=...)`
in `02_transform` so the `_build.json` manifest pins the source versions.
